## 2. 데이터 선정 이유

### 계기

지인에게 지역화폐로 자녀들의 학원비를 낸다는 이야기를 듣고 그렇다면 지역화폐는 소상공인, 지역상권에 도움을 주지 못하고 정기적인 소비를 대체하는것이 아닌가? 라는 의구심이 들었다. 이러한 궁금증을 바탕으로 지역화폐 제도의 실효성에 대해 비판적으로 분석해보고 실효성이 있다면 어떤 요인이 상관관계가 있는지, 없다면 정책의 개선에 도움을 줄 수 있는 인사이트를 발굴해보고자 한다.

### 고른 이유

경기데이터드림에 지역화폐 소비량이 행정동 단위로 존재했다. 서울시의 경우 구단위로 소비량을 확인할 수 있어 보다 많은 행을 확보하기 위해 경기도를 공간적 범위로 설정하였고 소비량에 영향을 미치는 생활인구, 유동인구 데이터를 수집했고 특정 매장에서만 사용할 수 있다는 지역화폐의 특징을 반영하여 지역화폐 사용 가능 매장 데이터를 수집하였다.
