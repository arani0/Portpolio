## local_live 컬럼명 동일하게 변경 (기준연월 -> 기준년월)

```python
local_live = local_live.rename(
    columns={"기준연월": "기준년월"}
)
```

```python
# 1. 모든 데이터의 행정동코드를 문자열로 통일
for data in [local_sales, local_live, local_move, local_store]:
    data["행정동코드"] = (
        data["행정동코드"]
        .astype("Int64")
        .astype("string")
        .str.zfill(10)
    )

# 2. 월별 자료의 기준년월을 정수형으로 통일
for data in [local_sales, local_live, local_move]:
    data["기준년월"] = data["기준년월"].astype("Int64")

# 3. 분석 기간으로 제한
local_sales = local_sales.loc[
    local_sales["기준년월"].between(202307, 202506)
].copy()

local_live = local_live.loc[
    local_live["기준년월"].between(202307, 202506)
].copy()

local_move = local_move.loc[
    local_move["기준년월"].between(202307, 202506)
].copy()

# 4. 점포 자료는 2023~2025년 자료 유지
local_store["년도"] = local_store["년도"].astype("Int64")

local_store = local_store.loc[
    local_store["년도"].between(2023, 2025)
].copy()
```

## 무결한 행정동 찾기

```python
# 1. 월별 자료에서 24개월 모두 존재하는 행정동 찾기
complete_dongs = {}

for name, data in [
    ("sales", local_sales),
    ("live", local_live),
    ("move", local_move)
]:
    month_count = data.groupby("행정동코드")["기준년월"].nunique()

    complete_dongs[name] = set(
        month_count[month_count == 24].index
    )

    print(f"{name}: 24개월 모두 존재하는 행정동 {len(complete_dongs[name])}개")


# 2. 점포 자료에서 2023·2024·2025년 모두 존재하는 행정동 찾기
year_count = local_store.groupby("행정동코드")["년도"].nunique()

complete_dongs["store"] = set(
    year_count[year_count == 3].index
)

print(f"store: 3개년 모두 존재하는 행정동 {len(complete_dongs['store'])}개")


# 3. 네 데이터의 공통 행정동
common_dongs = (
    complete_dongs["sales"]
    & complete_dongs["live"]
    & complete_dongs["move"]
    & complete_dongs["store"]
)

print(f"\n최종 공통 행정동: {len(common_dongs)}개")
print(f"예상 병합 후 행 수: {len(common_dongs) * 24:,}행")
```

## 행정동 결측이 많은 데이터셋 확인

```python
dong_check = pd.DataFrame({
    "행정동코드": sorted(complete_dongs["sales"])
})

dong_check["인구_24개월"] = dong_check["행정동코드"].isin(
    complete_dongs["live"]
)

dong_check["유동인구_24개월"] = dong_check["행정동코드"].isin(
    complete_dongs["move"]
)

dong_check["점포_3개년"] = dong_check["행정동코드"].isin(
    complete_dongs["store"]
)

# 조건 조합별 행정동 수
check_summary = (
    dong_check
    .groupby(["인구_24개월", "유동인구_24개월", "점포_3개년"])
    .size()
    .reset_index(name="행정동수")
)

display(check_summary)
```

- 지역화폐 매출, 생활인구, 유동인구, 업종 별 매장 수 모든 데이터셋을 활용하기 위해 260개의 행정동만 사용하기로 결정

## 데이터 셋 병합

```python
# 1. 공통 행정동 260개만 선택
sales = local_sales.loc[
    local_sales["행정동코드"].isin(common_dongs)
].copy()

live = local_live.loc[
    local_live["행정동코드"].isin(common_dongs)
].copy()

move = local_move.loc[
    local_move["행정동코드"].isin(common_dongs)
].copy()

store = local_store.loc[
    local_store["행정동코드"].isin(common_dongs)
].copy()


# 2. 점포 자료: 연도·행정동별 업종 개수를 가로로 펼치기
store_wide = store.pivot_table(
    index=["년도", "행정동코드"],
    columns="업종명(종목명)",
    values="개수",
    aggfunc="sum",
    fill_value=0
).add_prefix("점포수_").reset_index()

store_wide.columns.name = None


# 3. 매출 + 인구 자료
merged_df = sales.merge(
    live[["기준년월", "행정동코드", "행정구역", "총인구수"]],
    on=["기준년월", "행정동코드"],
    how="left",
    validate="one_to_one"
)


# 4. 유동인구 자료 추가
merged_df = merged_df.merge(
    move[["기준년월", "행정동코드", "월유동인구수"]],
    on=["기준년월", "행정동코드"],
    how="left",
    validate="one_to_one"
)


# 5. 연도 생성 후 점포 자료 추가
merged_df["년도"] = merged_df["기준년월"] // 100

merged_df = merged_df.merge(
    store_wide,
    on=["년도", "행정동코드"],
    how="left",
    validate="many_to_one"
)


# 6. 행정동·월 순서로 정렬
merged_df = merged_df.sort_values(
    ["행정동코드", "기준년월"]
).reset_index(drop=True)

# 7. 병합 결과 확인
print("행·열 수:", merged_df.shape)
print("행정동 수:", merged_df["행정동코드"].nunique())
print(
    "월·행정동 중복:",
    merged_df.duplicated(["기준년월", "행정동코드"]).sum()
)

display(merged_df.head())

# 결측치가 있는 컬럼만 표시
missing = merged_df.isna().sum()
display(missing[missing > 0])
```

## 중복열 정리

```python
print("점포 자료 컬럼:")
print(local_store.columns.tolist())

print("\n병합에 사용된 업종 종류 수:")
print(store["업종명(종목명)"].nunique())

print("\n전체 업종 목록:")
industry_list = (
    store["업종명(종목명)"]
    .dropna()
    .drop_duplicates()
    .sort_values()
)

print(industry_list.to_string(index=False))
```

```python
# 원본 업종명은 유지하고 정리용 컬럼 생성
def clean_industry(value):
    if pd.isna(value):
        return pd.NA

    # 공백 제거
    name = "".join(str(value).split())

    # 슬래시 앞뒤가 완전히 같은 경우에만 하나로 줄이기
    for i, char in enumerate(name):
        if char == "/" and name[:i] == name[i + 1:]:
            return name[:i]

    return name


store["업종명_정리"] = store["업종명(종목명)"].apply(clean_industry)

print("정리 전:", store["업종명(종목명)"].nunique())
print("정리 후:", store["업종명_정리"].nunique())
```

```python
industry_list = (
    store["업종명_정리"]
    .dropna()
    .drop_duplicates()
    .sort_values()
)

print(industry_list.to_string(index=False))
```

```python
# 팀 분류 결과가 반영된 엑셀 파일을 노트북과 같은 폴더에 저장하세요.
mapping_file = "지역화폐_가맹점_업종대분류_기준표_팀분류반영.xlsx"
industry_map = pd.read_excel(mapping_file, sheet_name="업종매핑")
cols = ["업종명_정리", "대분류", "분류상태"]
industry_map = industry_map[cols]

# 분류표 자체의 중복과 허용되지 않은 분류값을 확인합니다.
if not industry_map["업종명_정리"].is_unique:
    duplicated_names = industry_map.loc[
        industry_map["업종명_정리"].duplicated(keep=False),
        "업종명_정리"
    ].sort_values()
    display(duplicated_names)
    raise ValueError("분류표의 업종명_정리에 중복이 있습니다.")

category_order = [
    "소매", "숙박", "음식", "부동산", "과학·기술", "시설관리·임대",
    "교육", "보건의료", "예술·스포츠", "수리·개인", "기타"
]

unexpected_categories = sorted(
    set(industry_map["대분류"].dropna()) - set(category_order)
)
if unexpected_categories:
    raise ValueError(f"분류표에 허용되지 않은 대분류가 있습니다: {unexpected_categories}")


def clean_industry(value):
    if pd.isna(value):
        return pd.NA

    name = "".join(str(value).split())

    for i, char in enumerate(name):
        if char == "/" and name[:i] == name[i + 1:]:
            return name[:i]

    return name


# 공통 행정동으로 제한했던 store를 사용합니다. 원본 store는 보존합니다.
store_classified = store.drop(
    columns=["대분류", "분류상태"], errors="ignore"
).copy()

store_classified["업종명_정리"] = (
    store_classified["업종명(종목명)"].apply(clean_industry)
)

store_classified = store_classified.merge(
    industry_map,
    on="업종명_정리",
    how="left",
    validate="many_to_one"
)

unmapped = store_classified["대분류"].isna()

if unmapped.any():
    display(
        store_classified.loc[
            unmapped, ["업종명(종목명)", "업종명_정리"]
        ].drop_duplicates()
    )
    raise ValueError(
        "분류표에 없는 업종 또는 업종명 결측이 있습니다. 먼저 확인하세요."
    )

store_classified["개수"] = pd.to_numeric(
    store_classified["개수"], errors="raise"
)

assert store_classified[["년도", "행정동코드", "개수"]].notna().all().all()
assert (store_classified["개수"] >= 0).all()
assert (store_classified["개수"] % 1 == 0).all()

# 연도별·대분류별 실제 지역화폐 가맹점 수를 확인합니다.
category_counts = (
    store_classified.groupby(["년도", "대분류"])["개수"]
    .sum()
    .unstack(fill_value=0)
    .reindex(columns=category_order, fill_value=0)
)
display(category_counts)

# 10개 대분류와 기타를 각각 하나의 열로 집계합니다.
store_wide_major = store_classified.pivot_table(
    index=["년도", "행정동코드"],
    columns="대분류",
    values="개수",
    aggfunc="sum",
    fill_value=0,
    observed=True
).reindex(columns=category_order, fill_value=0)

# 연도·행정동별 집계 전후 총 가맹점 수가 같은지 확인합니다.
before = (
    store_classified.groupby(["년도", "행정동코드"])["개수"]
    .sum()
    .sort_index()
)
after = store_wide_major.sum(axis=1).sort_index()

assert before.index.equals(after.index)
assert (before.to_numpy() == after.to_numpy()).all()

store_wide_major = store_wide_major.add_prefix("가맹점수_").reset_index()
store_wide_major.columns.name = None

# 기존 세부 업종 열을 제거한 사본에 새 11개 분류 열을 붙입니다.
old_store_cols = [
    c for c in merged_df.columns
    if c.startswith("점포수_") or c.startswith("가맹점수_")
]

base = merged_df.drop(columns=old_store_cols).copy()
base["년도"] = base["기준년월"].astype("Int64") // 100

merged_df_major = base.merge(
    store_wide_major,
    on=["년도", "행정동코드"],
    how="left",
    validate="many_to_one",
    indicator="가맹점병합확인"
)

not_matched = ~merged_df_major["가맹점병합확인"].eq("both")
if not_matched.any():
    display(
        merged_df_major.loc[
            not_matched, ["년도", "행정동코드", "가맹점병합확인"]
        ].drop_duplicates()
    )
    raise ValueError("가맹점 자료와 연결되지 않은 연도·행정동이 있습니다.")

merged_df_major = merged_df_major.drop(columns="가맹점병합확인")

assert len(merged_df_major) == len(merged_df)
assert not merged_df_major.duplicated(["기준년월", "행정동코드"]).any()

print("대분류 병합 후 행·열:", merged_df_major.shape)
display(merged_df_major.head())
```
